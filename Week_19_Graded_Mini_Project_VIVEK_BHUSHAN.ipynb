```python
# ===================================================
# Week_19_Graded Mini Project_VIVEK-BHUSHAN
# Fashion-MNIST AE, VAE, cGAN, StyleGAN-lite AdaIN
# ===================================================

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader
from torchvision import datasets, transforms
import matplotlib.pyplot as plt
from tqdm.auto import tqdm

# -------------------------
# Task 1: Setup & Data
# -------------------------

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print("Using device:", device)

batch_size = 128

transform = transforms.Compose([
    transforms.ToTensor(),    # Fashion-MNIST: [1,28,28], values in [0,1]
])

train_ds = datasets.FashionMNIST(
    root='./data', train=True, download=True, transform=transform
)
test_ds = datasets.FashionMNIST(
    root='./data', train=False, download=True, transform=transform
)

train_loader = DataLoader(train_ds, batch_size=batch_size, shuffle=True)
test_loader = DataLoader(test_ds, batch_size=batch_size, shuffle=False)

def show_grid(images, nrows=6, ncols=6, title=None):
    """
    Display a grid of images: images can be [B,1,28,28] or [B,28,28].
    """
    imgs = images.detach().cpu()
    if imgs.ndim == 4:
        imgs = imgs.squeeze(1)
    fig, axes = plt.subplots(nrows, ncols, figsize=(ncols, nrows))
    for i, ax in enumerate(axes.flat):
        if i < imgs.size(0):
            ax.imshow(imgs[i], cmap='gray')
        ax.axis('off')
    if title:
        fig.suptitle(title)
    plt.tight_layout()
    plt.show()


# ============================================================
# Task 2: Denoising Autoencoder
# ============================================================

class DenoisingAE(nn.Module):
    def __init__(self, latent_dim=32):
        super().__init__()
        # Encoder
        self.enc = nn.Sequential(
            nn.Conv2d(1, 32, 3, stride=2, padding=1),  # [32,14,14]
            nn.ReLU(),
            nn.Conv2d(32, 64, 3, stride=2, padding=1), # [64,7,7]
            nn.ReLU(),
        )
        self.enc_fc = nn.Linear(64*7*7, latent_dim)
        # Decoder
        self.dec_fc = nn.Linear(latent_dim, 64*7*7)
        self.dec = nn.Sequential(
            nn.ConvTranspose2d(64, 32, 3, stride=2, padding=1, output_padding=1),
            nn.ReLU(),
            nn.ConvTranspose2d(32, 1, 3, stride=2, padding=1, output_padding=1),
            nn.Sigmoid(),  # 28x28 output in [0,1]
        )

    def encode(self, x):
        h = self.enc(x)
        h = h.view(x.size(0), -1)
        z = self.enc_fc(h)
        return z

    def decode(self, z):
        h = self.dec_fc(z)
        h = h.view(z.size(0), 64, 7, 7)
        x_hat = self.dec(h)
        return x_hat

    def forward(self, x):
        z = self.encode(x)
        x_hat = self.decode(z)
        return x_hat


def add_noise(x, sigma=0.3):
    noise = torch.randn_like(x) * sigma
    x_noisy = x + noise
    return torch.clamp(x_noisy, 0.0, 1.0)

ae = DenoisingAE(latent_dim=32).to(device)
opt_ae = torch.optim.Adam(ae.parameters(), lr=1e-3)
criterion_ae = nn.MSELoss()

num_epochs_ae = 10

for epoch in range(num_epochs_ae):
    ae.train()
    running_loss = 0.0
    for x, _ in tqdm(train_loader, desc=f"[AE] Epoch {epoch+1}/{num_epochs_ae}"):
        x = x.to(device)
        x_noisy = add_noise(x)

        x_hat = ae(x_noisy)
        loss = criterion_ae(x_hat, x)

        opt_ae.zero_grad()
        loss.backward()
        opt_ae.step()

        running_loss += loss.item() * x.size(0)

    print(f"[AE] Epoch {epoch+1}, loss={running_loss/len(train_ds):.4f}")

# Figure 1: clean vs noisy vs denoised
ae.eval()
x_test, _ = next(iter(test_loader))
x_test = x_test.to(device)[:36]
x_noisy = add_noise(x_test)
with torch.no_grad():
    x_denoised = ae(x_noisy)

show_grid(x_test, title="AE - Clean (Figure 1a)")
show_grid(x_noisy, title="AE - Noisy (Figure 1b)")
show_grid(x_denoised, title="AE - Denoised (Figure 1c)")


# ============================================================
# Task 3: Variational Autoencoder (VAE)
# ============================================================

class VAE(nn.Module):
    def __init__(self, latent_dim=32):
        super().__init__()
        # Encoder
        self.enc = nn.Sequential(
            nn.Conv2d(1, 32, 3, stride=2, padding=1),
            nn.ReLU(),
            nn.Conv2d(32, 64, 3, stride=2, padding=1),
            nn.ReLU(),
        )
        self.enc_fc_mu = nn.Linear(64*7*7, latent_dim)
        self.enc_fc_logvar = nn.Linear(64*7*7, latent_dim)
        # Decoder
        self.dec_fc = nn.Linear(latent_dim, 64*7*7)
        self.dec = nn.Sequential(
            nn.ConvTranspose2d(64, 32, 3, stride=2, padding=1, output_padding=1),
            nn.ReLU(),
            nn.ConvTranspose2d(32, 1, 3, stride=2, padding=1, output_padding=1),
            nn.Sigmoid(),
        )

    def encode(self, x):
        h = self.enc(x)
        h = h.view(x.size(0), -1)
        mu = self.enc_fc_mu(h)
        logvar = self.enc_fc_logvar(h)
        return mu, logvar

    def reparameterize(self, mu, logvar):
        std = torch.exp(0.5 * logvar)
        eps = torch.randn_like(std)
        return mu + std * eps  # z = μ + exp(0.5*logσ²)*ε

    def decode(self, z):
        h = self.dec_fc(z)
        h = h.view(z.size(0), 64, 7, 7)
        x_hat = self.dec(h)
        return x_hat

    def forward(self, x):
        mu, logvar = self.encode(x)
        z = self.reparameterize(mu, logvar)
        x_hat = self.decode(z)
        return x_hat, mu, logvar


def vae_loss(x_hat, x, mu, logvar):
    # BCE recon + KL regulariser (ELBO)
    bce = F.binary_cross_entropy(x_hat, x, reduction='sum') / x.size(0)
    kl = -0.5 * torch.sum(1 + logvar - mu.pow(2) - logvar.exp()) / x.size(0)
    return bce + kl, bce, kl

vae = VAE(latent_dim=32).to(device)
opt_vae = torch.optim.Adam(vae.parameters(), lr=2e-3)
num_epochs_vae = 12

for epoch in range(num_epochs_vae):
    vae.train()
    running_loss = 0.0
    for x, _ in tqdm(train_loader, desc=f"[VAE] Epoch {epoch+1}/{num_epochs_vae}"):
        x = x.to(device)
        x_hat, mu, logvar = vae(x)
        loss, bce, kl = vae_loss(x_hat, x, mu, logvar)

        opt_vae.zero_grad()
        loss.backward()
        opt_vae.step()

        running_loss += loss.item() * x.size(0)

    print(f"[VAE] Epoch {epoch+1}, loss={running_loss/len(train_ds):.4f}")

# VAE Figures: recon, samples, interpolations, traversals
vae.eval()
with torch.no_grad():
    # Reconstructions
    x_test, _ = next(iter(test_loader))
    x_test = x_test.to(device)[:36]
    x_hat, mu, logvar = vae(x_test)
    show_grid(x_test, title="VAE - Original (Recon)")
    show_grid(x_hat, title="VAE - Reconstruction (Figure 2a)")

    # Random samples (Figure 2b)
    z_rand = torch.randn(36, 32, device=device)
    samples = vae.decode(z_rand)
    show_grid(samples, title="VAE - Random Samples (Figure 2b)")


def interpolate(z1, z2, steps=10):
    alphas = torch.linspace(0, 1, steps).unsqueeze(1).to(z1.device)
    return (1 - alphas) * z1 + alphas * z2

vae.eval()
with torch.no_grad():
    # Interpolations (Figure 3)
    z1 = torch.randn(1, 32, device=device)
    z2 = torch.randn(1, 32, device=device)
    z_interp = interpolate(z1, z2, steps=10)
    x_interp = vae.decode(z_interp)
    show_grid(x_interp, nrows=2, ncols=5, title="VAE - Interpolations (Figure 3)")

    # Traversals (Figure 4)
    z_base = torch.randn(1, 32, device=device)
    z_traversals = []
    vals = torch.linspace(-3, 3, steps=10)
    for v in vals:
        z = z_base.clone()
        z[0, 0] = v
        z_traversals.append(z)
    z_traversals = torch.cat(z_traversals, dim=0)
    x_trav = vae.decode(z_traversals)
    show_grid(x_trav, nrows=2, ncols=5, title="VAE - Traversal dim0 (Figure 4)")


# ============================================================
# Task 4: Conditional GAN (cGAN)
# ============================================================

num_classes = 10
latent_dim = 64
embed_dim = 10  # label embedding size

class Generator(nn.Module):
    def __init__(self, latent_dim, num_classes, embed_dim):
        super().__init__()
        self.embed = nn.Embedding(num_classes, embed_dim)
        input_dim = latent_dim + embed_dim
        self.fc = nn.Linear(input_dim, 256*7*7)
        self.net = nn.Sequential(
            nn.ConvTranspose2d(256, 128, 4, stride=2, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(),
            nn.ConvTranspose2d(128, 64, 4, stride=2, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(),
            nn.Conv2d(64, 1, 3, padding=1),
            nn.Sigmoid(),
        )

    def forward(self, z, labels):
        y = self.embed(labels)             # [B, embed_dim]
        x = torch.cat([z, y], dim=1)       # [B, latent_dim+embed_dim]
        x = self.fc(x)
        x = x.view(-1, 256, 7, 7)
        img = self.net(x)
        return img


class Discriminator(nn.Module):
    def __init__(self, num_classes, embed_dim):
        super().__init__()
        self.embed = nn.Embedding(num_classes, embed_dim)
        self.label_fc = nn.Linear(embed_dim, 28*28)
        self.net = nn.Sequential(
            nn.Conv2d(2, 64, 4, stride=2, padding=1),
            nn.LeakyReLU(0.2),
            nn.Conv2d(64, 128, 4, stride=2, padding=1),
            nn.BatchNorm2d(128),
            nn.LeakyReLU(0.2),
            nn.Flatten(),
            nn.Linear(128*7*7, 1),
        )

    def forward(self, img, labels):
        y = self.embed(labels)
        y_map = self.label_fc(y).view(-1, 1, 28, 28)
        x = torch.cat([img, y_map], dim=1)
        logits = self.net(x)
        return logits


def sample_z(batch_size, latent_dim):
    return torch.randn(batch_size, latent_dim, device=device)


G = Generator(latent_dim, num_classes, embed_dim).to(device)
D = Discriminator(num_classes, embed_dim).to(device)

criterion_gan = nn.BCEWithLogitsLoss()
opt_G = torch.optim.Adam(G.parameters(), lr=2e-4, betas=(0.5, 0.999))
opt_D = torch.optim.Adam(D.parameters(), lr=2e-4, betas=(0.5, 0.999))

num_epochs_cgan = 12  # 10–15

for epoch in range(num_epochs_cgan):
    for real_imgs, labels in tqdm(train_loader, desc=f"[cGAN] Epoch {epoch+1}/{num_epochs_cgan}"):
        real_imgs = real_imgs.to(device)
        labels = labels.to(device)
        bsz = real_imgs.size(0)

        # --- Train D ---
        z = sample_z(bsz, latent_dim)
        fake_labels = labels
        fake_imgs = G(z, fake_labels).detach()

        real_logits = D(real_imgs, labels)
        fake_logits = D(fake_imgs, fake_labels)

        real_targets = torch.full_like(real_logits, 0.9)  # label smoothing
        fake_targets = torch.zeros_like(fake_logits)

        loss_D_real = criterion_gan(real_logits, real_targets)
        loss_D_fake = criterion_gan(fake_logits, fake_targets)
        loss_D = loss_D_real + loss_D_fake

        opt_D.zero_grad()
        loss_D.backward()
        opt_D.step()

        # --- Train G ---
        z = sample_z(bsz, latent_dim)
        gen_labels = labels
        gen_imgs = G(z, gen_labels)
        gen_logits = D(gen_imgs, gen_labels)

        gen_targets = torch.ones_like(gen_logits)
        loss_G = criterion_gan(gen_logits, gen_targets)

        opt_G.zero_grad()
        loss_G.backward()
        opt_G.step()

    print(f"[cGAN] Epoch {epoch+1}, D_loss={loss_D.item():.4f}, G_loss={loss_G.item():.4f}")

    # Figure 5: one sample per class 0–9
    G.eval()
    with torch.no_grad():
        labels_grid = torch.arange(0, 10, device=device)
        z_grid = sample_z(10, latent_dim)
        imgs_grid = G(z_grid, labels_grid)
    show_grid(imgs_grid, nrows=2, ncols=5, title=f"cGAN - Class Samples (Figure 5, epoch {epoch+1})")
    G.train()

# Optional Figure 6: diversity grid
G.eval()
with torch.no_grad():
    labels_div = torch.repeat_interleave(torch.arange(0, 10, device=device), 4)  # 4 per class
    z_div = sample_z(labels_div.size(0), latent_dim)
    imgs_div = G(z_div, labels_div)
show_grid(imgs_div, nrows=4, ncols=10, title="cGAN - Diversity Grid (Figure 6)")
G.train()


# ============================================================
# Task 5: StyleGAN-lite AdaIN Demo
# ============================================================

feature_extractor = nn.Sequential(
    nn.Conv2d(1, 32, 3, padding=1),
    nn.ReLU(),
    nn.Conv2d(32, 64, 3, padding=1),
    nn.ReLU(),
).to(device)

decoder_style = nn.Sequential(
    nn.Conv2d(64, 32, 3, padding=1),
    nn.ReLU(),
    nn.Conv2d(32, 1, 3, padding=1),
    nn.Sigmoid(),
).to(device)

def adain(content, style, eps=1e-5):
    """
    Adaptive Instance Normalisation:
    normalize content, then apply style mean/std.
    """
    c_mean = content.mean(dim=[2, 3], keepdim=True)
    c_std = content.std(dim=[2, 3], keepdim=True) + eps
    s_mean = style.mean(dim=[2, 3], keepdim=True)
    s_std = style.std(dim=[2, 3], keepdim=True) + eps
    normalized = (content - c_mean) / c_std
    return normalized * s_std + s_mean

# Use 8 content and 8 style images
with torch.no_grad():
    x_content, _ = next(iter(test_loader))
    x_style, _ = next(iter(train_loader))
    x_content = x_content.to(device)[:8]
    x_style = x_style.to(device)[:8]

    f_content = feature_extractor(x_content)
    f_style = feature_extractor(x_style)

    f_mixed = adain(f_content, f_style)
    x_mixed = decoder_style(f_mixed)

show_grid(x_content, nrows=2, ncols=4, title="AdaIN - Content (Figure 7a)")
show_grid(x_style, nrows=2, ncols=4, title="AdaIN - Style (Figure 7b)")
show_grid(x_mixed, nrows=2, ncols=4, title="AdaIN - Mixed (Figure 7c)")

# Thank You for evaluating Week_19_Graded_Mini_Project_VIVEK_BHUSHAN
```
